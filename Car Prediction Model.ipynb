```python
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder,StandardScaler,FunctionTransformer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,RidgeCV,LassoCV
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
print(df.info())
```

```python
x=df.drop(columns='Price')
y=df['Price']

xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=42)
print(xtrain.head())
print(ytrain)
```

```python
cat_col=['Name','Location','Fuel_Type','Transmission']

ordi_col=['Owner_type']

std_sclr_col=['Year','Seats','Mileage']

ordi_col=['Owner_Type']

std_sclr_col=['Year','Seats']

log_std_sclr_col=['Kilometers_Driven','Engine','Power']

pipe1=OneHotEncoder(handle_unknown='ignore')

pipe2=OrdinalEncoder(categories=[['First', 'Second', 'Third', 'Fourth & Above']]
)

pipe3=StandardScaler()

pipe4=Pipeline([
    ('funtrsf',FunctionTransformer(np.log1p)),
    ('stdsclr',StandardScaler())
])

pre=ColumnTransformer([
    ('c1',pipe1,cat_col),
    ('c2',pipe2,ordi_col),
    ('c3',pipe3,std_sclr_col),
    ('c4',pipe4,log_std_sclr_col)
])

model=Pipeline([
    ('preprocessor',pre),
    ('lrmodel',LinearRegression())
])

ytrainnew=np.log1p(ytrain)
ytestnew=np.log1p(ytest)

model.fit(xtrain,ytrainnew)

ypred=model.predict(xtest)
ypred_actual=np.expm1(ypred)

print(ypred_actual)
print(ytest)

print(r2_score(ytest,ypred_actual))
print(mean_absolute_error(ytest,ypred_actual))
print(mean_squared_error(ytest,ypred_actual))
```

```python
model2=Pipeline([
    ('preprocessor',pre),
    ('lrmodel',RidgeCV(cv=5))
])

ytrainnew_ridge=np.log1p(ytrain)
ytestnew_ridge=np.log1p(ytest)

model2.fit(xtrain,ytrainnew_ridge)

ypred_ridge=model2.predict(xtest)
ypred_actual_ridge=np.expm1(ypred_ridge)

print(ypred_actual_ridge)
print(ytest)

print(r2_score(ytest,ypred_actual_ridge))
print(mean_absolute_error(ytest,ypred_actual_ridge))
print(mean_squared_error(ytest,ypred_actual_ridge))
print(model2.named_steps['lrmodel'].alpha_)
```

```python
model3=Pipeline([
    ('preprocessor',pre),
    ('lrmodel',LassoCV(cv=5))
])

ytrainnew_lasso=np.log1p(ytrain)
ytestnew_lasso=np.log1p(ytest)

model3.fit(xtrain,ytrainnew_lasso)

ypred_lasso=model3.predict(xtest)
ypred_actual_lasso=np.expm1(ypred_lasso)

print(ypred_actual_lasso)
print(ytest)

print(r2_score(ytest,ypred_actual_lasso))
print(mean_absolute_error(ytest,ypred_actual_lasso))
print(mean_squared_error(ytest,ypred_actual_lasso))
print(model3.named_steps['lrmodel'].alpha_)
```
